METODO CERRADO / METODO BISECCION

Código con ciclo for

```python
import numpy as np

def f(x):
  return x - np.cos(x)

lista_c = [ ]
a = -4
b = 3

for i in range(10):
  c = (a+b)/ 2 # conseguir la biseccion
  lista_c.append(c) #Agregar la bisección a la lista

  f_a = f(a)
  f_c = f(c)

  if f_a*f_c < 0:
    b = c
  else:
    a = c
print(lista_c)
```

Ciclo while. Calculando cifras de precisión (REVISAR!!!)

```python
import numpy as np

def f(x):
    return x - np.cos(x)

a = -4
b = 3

lista_c = []

es = 0.0005      # Error permitido para 5 cifras significativas
er = 1           # Se inicia con un valor grande
c_anterior = 0
iteraciones=0

while er > es:

    c = (a + b) / 2
    lista_c.append(c)

    # Calcular error (excepto en la primera iteración)
    if len(lista_c) > 1:
        er = abs((c - c_anterior) / c)

    c_anterior = c

    # Método de bisección
    if f(a) * f(c) < 0:
        b = c
    else:
        a = c
    iteraciones = iteraciones + 1
print("Raíz aproximada:", c)
print("Error relativo:", er)
print("Número de iteraciones:", iteraciones)
print(lista_c)
```

METODO CERRADO / Metodo falsa Posicion (REVISAR!!!)

```python
import numpy as np

def f(x):
    return x - np.cos(x)

lista_c = [ ]

a = -4
b = 3

for i in range(10):

    f_a = f(a)
    f_b = f(b)

    # Fórmula de falsa posición
    c = b - (f_b * (a - b)) / (f_a - f_b)

    lista_c.append(float(c))

    f_c = f(c)

    if f_a * f_c < 0:
        b = c
    else:
        a = c

print(lista_c)
```
