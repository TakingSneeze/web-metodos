#Regla del trapecio

Se aproxima la integral como- una función lineal (b-a) (f(a)+f(b))/2

**trapecio multiple:** El ancho del trapecio de modifica:

    ancho = (b-a)/# de trapecios



```python
import numpy as np
import matplotlib.pyplot as plt
def f(x):
  return np.sin(x) + 2

a = 1
b = 6
trapecios = 100
puntos_x = trapecios + 1

x = np.linspace(a,b,puntos_x)
y = f(x)

#sumas de areas
area_total=0
#plt.plot(x,y)

#usamos un ciclo for que nos calcule el area de cada trapecio
for i in range(4):
  base = x[i+1] - x[i]
  altura = (y[i+1] + y[i])/2
  area_trapecio = base * altura
  area_total = area_total + area_trapecio

  print(area_total)

  #PARA PONERLO EN UN WHILE PARA EL PARCIAL:
  #While No.Trapecios aumenta, el error = abs(antrap-antrap-1)/antrap donde er=0.5*10**-n
```

Se puede aproximar la integral como una cuadrática (ax**2 + bx + c)

siendo h la mitad de la longitud del trapecio.

```python
import numpy as np
import matplotlib.pyplot as plt
def f(x):
  return np.sin(x) + 2

a = 1
b = 6
polinomios = 10
puntos_x = polinomios*2 + 1

x = np.linspace(a,b,puntos_x)
y = f(x)
print(x)
#sumas de areas
area_total=0
#plt.plot(x,y)

#usamos un ciclo for que nos calcule el area de cada trapecio
for i in range(polinomios):
  base = [i*2+2] - x[i*2]
  altura = (y[i*2] + 4*y[i*2+1]+y[i*2 + 2])/6
  area_trapecio = base * altura
  area_total = area_total + area_trapecio

  print(area_total)
```

```python
import numpy as np
import matplotlib.pyplot as plt
def f(x):
  return np.sin(x) + 2

a = 1
b = 6
polinomios = 10
puntos_x = polinomios*2 + 1

x = np.linspace(a,b,puntos_x)
y = f(x)
print(x)
#sumas de areas
area_total=0
#plt.plot(x,y)

#usamos un ciclo for que nos calcule el area de cada trapecio
for i in range(polinomios):
  base = [i*3+2] - x[i*3]
  altura = (y[i*3] + 3*y[i-3+1]+3*y[i*3 + 3])/8
  area_trapecio = base * altura
  area_total = area_total + area_trapecio

  print(area_total)
```
