#Momento Flexionante

  Se tiene una viga de 11m de longitud, donde su esfuerzo cortante está dado por la función:
  
    5 + 0.25*(x**2)

  Calcular el momento flectante sabiendo que el esfuerzo cortante es la taza de cambio del momento flexionante con respecto a x

```python
import numpy as np
def v(x):
   return 5 + 0.25*(x**2)

def m(x):
  return 5*x + (0.25*(x**3))/3

a = 0
b = 11
n = 10

h = (b-a)/2

areatotaltotal= 0

#sumar area trapecios
area_totalt = 0
#sumar area polinomios
area_totalp = 0
#integral evaluada
for i in range(12):
  momento = m(i)

#cifras error
cifras = 0.5 * 10**-5



#trapecios
trap = n
puntos_xt= np.linspace(a,h,trap+1)
puntos_yt = v(puntos_xt)

for i in range (trap):
 base = puntos_xt[i+1] - puntos_xt[i]
 altura = (puntos_yt[i+1] + puntos_yt[i])/2
 areat = base * altura
 area_totalt += areat

#Simpson

poli = n

polirangos = poli*2 + 1
puntos_xp = np.linspace(h,b,polirangos)
puntos_yp = v(puntos_xp)

hsimpson = puntos_xp[1] - puntos_xp[0]

for j in range (poli):
  areapoli = (hsimpson / 3) * (puntos_yp[j*2] + 4*puntos_yp[j*2+1] + puntos_yp[j*2 + 2])
  area_totalp = area_totalp + areapoli

areatotaltotal = area_totalp + area_totalt

#error relativo
e_r = abs((momento-areatotaltotal)/momento)*100
#print(f'puntos en x: {puntos_x}')
print("")
print(f'Area por trapecios: {area_totalt}')
print(f'Area por polinomios: {area_totalp}')
print(f'Area total: {areatotaltotal}')
print("")
print(f'Area por integral normal: {momento}')
print("")
print(f'Error relativo: {e_r} %')
```
