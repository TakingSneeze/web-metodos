1b. Falsa posición

```python
import numpy as np

Q = 20
g = 9.81

x1 = 0.5
x2 = 2.5

error = 100
iteraciones = 0

lista_x = []
lista_error = []

while error >= 1:

    # B y Ac para x1
    B1 = 3 + x1
    Ac1 = 3*x1 + x1**2/2

    # B y Ac para x2
    B2 = 3 + x2
    Ac2 = 3*x2 + x2**2/2

    # Funciones
    f1 = 1 - (Q**2*B1)/(g*Ac1**3)
    f2 = 1 - (Q**2*B2)/(g*Ac2**3)

    # Falsa posición
    x3 = x2 - f2*(x2-x1)/(f2-f1)

    # Calcular función en x3
    B3 = 3 + x3
    Ac3 = 3*x3 + x3**2/2

    f3 = 1 - (Q**2*B3)/(g*Ac3**3)

    if iteraciones == 0:
        error = 100
    else:
        error = abs((x3-x_anterior)/x3)*100

    iteraciones = iteraciones + 1

    lista_x.append(x3)
    lista_error.append(error)

    if f1*f3 < 0:
        x2 = x3
    else:
        x1 = x3

    x_anterior = x3

print("Profundidad crítica =", x3)
print("Iteraciones =", iteraciones)
```

2. Diseño de un tanque

Newton-Raphson

```python
R = 3
V = 30

h = R

iteraciones = 0

while iteraciones < 3:

    # Función
    f = np.pi*h**2*(3*R-h)/3 - V

    # Derivada
    df = np.pi*(6*h-h**2)

    # Newton-Raphson
    h_nuevo = h - f/df

    # Error
    error = abs((h_nuevo-h)/h_nuevo)*100

    iteraciones = iteraciones + 1

    h = h_nuevo
    print("Iteración:", iteraciones)
    print("h =", h_nuevo)
    print("Error =", error)
    print()

```

3.a Métodos de integración Trapecios

```python
import numpy as np

import numpy as np

g = 9.8
m = 68.1
cd = 0.25

a = 0
b = 10

n = 212

h = (b-a)/n

suma = 0

i = 1

while i < n:

    t = a + i*h

    v = np.sqrt(g*m/cd)*np.tanh(np.sqrt(g*cd/m)*t)

    suma = suma + v

    i = i + 1

v_a = np.sqrt(g*m/cd)*np.tanh(np.sqrt(g*cd/m)*a)
v_b = np.sqrt(g*m/cd)*np.tanh(np.sqrt(g*cd/m)*b)

distancia = h*((v_a+v_b)/2+suma)

print("Distancia =", distancia)
print("n =", n)
```

```python
import numpy as np

g = 9.8
m = 68.1
cd = 0.25

def v(t):
    return np.sqrt(g*m/cd)*np.tanh(np.sqrt(g*cd/m)*t)

def trapecio(n):
    a = 0
    b = 10
    h = (b-a)/n

    suma = 0

    for i in range(1,n):
        t = a + i*h
        suma += v(t)

    return h*((v(a)+v(b))/2 + suma)

n = 2
I_anterior = trapecio(n)

while True:
    n = n*2

    I_actual = trapecio(n)

    error = abs((I_actual-I_anterior)/I_actual)*100

    if error < 0.000005:
        break

    I_anterior = I_actual

print("n =", n)
print("Distancia =", I_actual)
print("Error =", error,"%")
```

3.b Método Simpson 1/3

```python
import numpy as np

g = 9.8
m = 68.1
cd = 0.25

a = 0
b = 10

n = 14

h = (b-a)/n

suma = 0

i = 1

while i < n:

    t = a + i*h

    v = np.sqrt(g*m/cd)*np.tanh(np.sqrt(g*cd/m)*t)

    if i % 2 == 0:
        suma = suma + 2*v
    else:
        suma = suma + 4*v

    i = i + 1

v_a = np.sqrt(g*m/cd)*np.tanh(np.sqrt(g*cd/m)*a)
v_b = np.sqrt(g*m/cd)*np.tanh(np.sqrt(g*cd/m)*b)

distancia = h/3*(v_a+v_b+suma)

print("Distancia =", distancia)
print("n =", n)
```

```python
import numpy as np

a = 0
b = 5.5

n = 92

h = (b-a)/n

suma = 0

i = 1

while i < n:

    x = a + i*h

    V = 5 + 0.25*x**2

    suma = suma + V

    i = i + 1

V_a = 5 + 0.25*a**2
V_b = 5 + 0.25*b**2

M_trap = h*((V_a+V_b)/2+suma)
```
