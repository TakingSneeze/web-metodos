Programa que vaya calculando cada valor de la función 1/x, empezando en x=1 con un avance en x de 0.0001. Se debe detener cuando el error relativo sea el equivalente a una precisión de 4 cifras significativas. ¿Cuántas iteraciones se requirieron?

```python
x=1
avance = 0.0001
y_actual = 1/x
error = 1
error_deseado = 0.5*10**4
iteraciones = 0
while error > error_deseado:
  x = x + avance
  y_ant=y_actual
  y_actual = 1/x
  error = abs((y_actual - y_ant)/y_ant)
  iteraciones = iteraciones + 1

print (x)
print (error)
print (iteraciones)
```
