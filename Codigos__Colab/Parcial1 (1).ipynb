**Primer punto**

Graficar las primeras 20 aproximaciones de x para encontrar el valor que satisface la ecuación

     2 - 2*x = -x**3

```python
import matplotlib.pyplot as plt
import numpy as np

def f(x):
    return x**3 - 2*x + 2
def derivadaf(x):
    return 3*x**2 - 2

x = 1

aproximaciones_enx = [x]
n = 0

#Usamos el método Newthon Rhapson
while n < 20:
    x = x - f(x)/derivadaf(x)
    aproximaciones_enx.append(x)
    n += 1

print(aproximaciones_enx)

#Grafica de aproximaciones
plt.plot(range(len(aproximaciones_enx)), aproximaciones_enx, marker='o')

plt.xlabel("No. de iteraciones")
plt.ylabel("Valores x")

plt.grid()
plt.show()

#Comprobar reiz real
x2 = -2
n = 0

while n < 20:
    x2 = x2 - f(x2)/derivadaf(x2)
    n += 1

print("")
print("Comprobación de raíz:")
print(f"Valor raíz real = {x2}")
```

*Conclusión del método utilizado:*

Si usamos el método de Newthon-Rhapson, No siempre va a converger, porque en esta función si tenemos una raíz (-1.769) pero como el método se evalúa en un x = 1 eso hace que se genere un bucle entre ceros y unos como podemos ver en la gráfica.


#Segundo Punto
Calcular la integral de la función sinuosoidal entre [pi, 2*pi] usando  **30** trapecios.

```python
import numpy as np

#La función va de 5(maximo) a -5(minimo) = 5 (-sin(x))
def f(x):
    return -5*np.sin(x)

a = np.pi
b = 2*np.pi
#Trapecios
n = 30

#Tamaño de los trapecios
h = (b-a)/n

puntos_x = np.linspace(a, b, n+1)
puntos_y = f(puntos_x)

area_total = 0

for i in range(n):
    base = puntos_x[i+1] - puntos_x[i]
    altura = (puntos_y[i+1] + puntos_y[i])/2
    area = base * altura
    area_total += area

print(f"Área total aproximada = {area_total}")

#Área real
def integralf(x):
  return 5*np.cos(x)

area_real = integralf(b) - integralf(a)

print(f"Área real = {area_real}")

error_relativo = abs((area_real - area_total)/area_real)*100
print(f"Error relativo = {error_relativo}%")
```
