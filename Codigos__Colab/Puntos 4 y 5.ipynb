```python
import numpy as np
def v(x):
   return 5 + 0.25*(x**2)

def m(x):
  return 5*x + (0.25*(x**3))/3

a = 0
b = 11
n = 4
e_r = 100

h = (a+b)/2

#cifras error
cifras = 0.0005

#integral evaluada
for i in range(12):
  momento = m(i)

while e_r > cifras :
  #trapecios
  trap = n
  puntos_xt= np.linspace(a,h,trap+1)
  puntos_yt = v(puntos_xt)

  area_totalt = 0
  i=0

  while i < n:
    base = puntos_xt[i+1] - puntos_xt[i]
    altura = (puntos_yt[i+1] + puntos_yt[i])/2
    areat = base * altura
    area_totalt += areat
    i += 1

  #Simpson

  polirangos = n*2 + 1
  puntos_xp = np.linspace(h,b,polirangos)
  puntos_yp = v(puntos_xp)

  area_totalp = 0
  hsimpson = puntos_xp[1] - puntos_xp[0]

  j = 0

  while j < n:
    areapoli = (hsimpson / 3) * (puntos_yp[j*2] + 4*puntos_yp[j*2+1] + puntos_yp[j*2 + 2])
    area_totalp = area_totalp + areapoli
    j +=1

  #Area total calculada
  areatotaltotal = area_totalp + area_totalt

  #error relativo
  e_r = abs((momento-areatotaltotal)/momento)*100

  n +=1


#print(f'puntos en x: {puntos_x}')
print(f"iteraciones: {n-4}")
print(f'Area por trapecios: {area_totalt}')
print(f'Area por polinomios: {area_totalp}')
print(f'Area total: {areatotaltotal}')
print("")
print(f'Area por integral normal: {momento}')
print("")
print(f'Error relativo: {e_r} %')
```
