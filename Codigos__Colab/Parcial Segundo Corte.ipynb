Se desea construir un acuario rectangular sin tapa de vidrio que deba contener exactamente $32\text{ m}^3$ de agua. ¿Cuáles deben ser las dimensiones de la base y la altura para minimizar la cantidad de vidrio utilizado? Use el método de la sección dorada multivariado con una precisión de 6 cifras en el area de vidrio usado.

```python
import numpy as np

#funcion del area del acuario
def area(x, y):
    return x*y + 64/x + 64/y

#metodo de la seccion dorada
def seccion_dorada(funcion, a, b, tol=1e-8):
    # factor d
    d = (np.sqrt(5) - 1)/2
    c = b - d*(b-a)
    e = a + d*(b-a)

    while abs(b-a) > tol:
        if funcion(c) < funcion(e):
            b = e
            e = c
            c = b - d*(b-a)
        else:
            a = c
            c = e
            e = a + d*(b-a)
    return (a+b)/2

#Optimizacion
x = 4
y = 4
area_anterior = area(x, y)

error = 1
iteracion = 0

while error > 1e-6:
    funcion_x = lambda x: area(x, y)
    x = seccion_dorada(funcion_x, 1, 10)

    funcion_y = lambda y: area(x, y)
    y = seccion_dorada(funcion_y, 1, 10)

    area_nueva = area(x, y)
# Error
    error = abs(area_nueva - area_anterior)
    area_anterior = area_nueva

    iteracion += 1

h = 32/(x*y)

print("numero de iteraciones=", iteracion)
print("largo x=", x, "m")
print("ancho y=", y, "m")
print("altura h=", h, "m")
print("area minima=", area_nueva, "m^2")
print("error=", error)
```
